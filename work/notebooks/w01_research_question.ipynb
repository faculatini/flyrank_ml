## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research question

**Question**

Which pages should be prioritized for editorial review as potential refresh opportunities?

**Decision**

An editor needs to decide which pages to review first.

**Action**

Review the highest-priority pages and decide whether to refresh, expand, monitor, or leave them unchanged.

**Cost of a wrong recommendation**

A false positive wastes editorial time by recommending pages that do not actually need attention.

A false negative is potentially more expensive because an important declining page may continue losing visibility while it is ignored.

**Why ML can help**

Manually reviewing thousands of pages is impractical. Machine learning can identify patterns across multiple signals and rank pages by priority, helping editors allocate their limited time more effectively.
